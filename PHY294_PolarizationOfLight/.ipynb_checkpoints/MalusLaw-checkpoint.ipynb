# Polarization of Light --> Malus Law

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

## Exercise 1: Two Polarizers
Export data, analyze dependencies: Intensity vs. cos θ, and Intensity vs. cos<sup>2</sup>θ

### Organize data

```python
# import data
ex1_01 = pd.read_csv("./Exercise1/Ex1_01.txt", sep='\t', header=1, skiprows=0)
ex1_02 = pd.read_csv("./Exercise1/Ex1_02.txt", sep='\t', header=1, skiprows=0)
ex1_03 = pd.read_csv("./Exercise1/Ex1_03.txt", sep='\t', header=1, skiprows=0)

display(ex1_01.head())
display(ex1_02.tail())

ex1 = [ex1_01, ex1_02, ex1_03]
display(ex1[2].tail())
```

```python
# order by intensity 
for i in range(0, len(ex1)):
    ex1[i] = ex1[i].sort_values(by=["Light Intensity (Volts)"], ascending=True)
    ex1[i] = ex1[i].reset_index(drop=True)
display(ex1[2])
```

### Plot Intensity vs cos(θ)

```python
# centred quadratic
# x is cos(θ), a is I_1
def cq(x, a):
    return a * x**2
```

```python
# save fit parameters for later
a = [0] * len(ex1)

for i in range(0, len(ex1)):
    # make the data
    x = np.array(np.cos(ex1[i]['Sensor Position (Radians)'].to_list())) # cos(θ)
    y = np.array(ex1[i]['Light Intensity (Volts)'].to_list()) # intensity
    
    # MAKE ERRORBARS
    # if y = acos(b*x)
    # y_unc = -(x_unc)*a*b*sin(b*x)
    x_err = 0.00001
    y_err = np.array(np.abs(-x_err*np.sin(ex1[i]['Sensor Position (Radians)'].to_list())))
    
    # plot
    fig, ax = plt.subplots()
    ax.scatter(x, y, label="data")
    ax.set_ylabel("Light Intensity (V)", fontsize=15)
    ax.set_xlabel("cos(θ)", fontsize=15)
    ax.errorbar(x, y, xerr=x_err, yerr=y_err, fmt='none')
    fig.set_size_inches(10, 7)

    # fit
    popt, pcov = curve_fit(cq, x, y, p0 = (4))
    # sort so the plotted line is correct
    x_new = np.sort(x)
    ax.plot(x_new, cq(x_new, *popt), label='Malus\' Law (quadratic fit)', color='orange')
    print(popt)
    a[i] = popt[0]
    
    # save
    ax.legend()
    name = "./intcos/intensityvscos_" + str(i) + ".png"
    fig.savefig(name, bbox_inches='tight', pad_inches=0.2)
```

```python
# RESIDUALS
for i in range(0, len(ex1)):

    fig, ax = plt.subplots()

    # sort df so y aligns with fit data
    df = ex1[i].sort_values(by=['Sensor Position (Radians)'])
    
    # make the data
    x = np.array(np.cos(df['Sensor Position (Radians)'].to_list())) # cos(θ)
    y = np.array(df['Light Intensity (Volts)'].to_list()) # intensity
    # errorbars
    # if y = acos(b*x)
    # y_unc = -(x_unc)*a*b*sin(b*x)
    x_err = 0.00001
    y_err = np.array(np.abs(-x_err*np.sin(df['Sensor Position (Radians)'].to_list())))

    # horizontal line
    ax.axhline(y=0, color="orange")
    # curve of best fit
    fit = a[i]*x**2
    res = y-fit
    
    # plot
    ax.scatter(x, res)
    ax.set_xlabel("cos(θ)", fontsize=15)
    ax.set_ylabel("Light Intensity Residuals (V)", fontsize=15)
    ax.errorbar(x, res, xerr=x_err, yerr=y_err, fmt='none')
    fig.set_size_inches(10, 7)

    # save
    #ax.legend()
    name = "./intcos/intcos_res_" + str(i) + ".png"
    fig.savefig(name, bbox_inches='tight', pad_inches=0.2)

print(a)
```

### Plot Intensity vs cos²(θ)

```python
# where x is cos²(θ)
# a is I_1
def malus(x, a):
    return a * x
```

```python
for i in range(0, len(ex1)):
    # make the data
    x = np.array((np.cos(ex1[i]['Sensor Position (Radians)'].to_list()))**2) # cos(θ)
    y = np.array(ex1[i]['Light Intensity (Volts)'].to_list()) # intensity
    
    # errorbars
    x_err = 0.00001
    y_err = -np.sin(2*x)*x_err
    #print(y_err[0:5])
    
    # plot
    fig, ax = plt.subplots()
    ax.scatter(x, y, label='data')
    ax.set_ylabel("Light Intensity (V)", fontsize=15)
    ax.set_xlabel("cos²(θ)", fontsize=15)
    #ax.errorbar(x, y, xerr=x_err, yerr=y_err[i], fmt='none')
    fig.set_size_inches(10, 7)

    # fit
    popt, pcov = curve_fit(malus, x, y, p0 = (4))
    ax.plot(x, malus(x, *popt), label='Malus\' Law (linear fit)', color='orange')
    a[i] = popt[0] # save for later
    print(popt)
    
    # save
    ax.legend()
    name = "./intcos2/intensityvscos2_" + str(i) + ".png"
    fig.savefig(name, bbox_inches='tight', pad_inches=0.5)
```

```python
# RESIDUALS
for i in range(0, len(ex1)):

    fig, ax = plt.subplots()

    # sort df so y aligns with fit data
    df = ex1[i].sort_values(by=['Sensor Position (Radians)'])
    
    # make the data
    x = np.array((np.cos(ex1[i]['Sensor Position (Radians)'].to_list()))**2) # cos(θ)
    y = np.array(ex1[i]['Light Intensity (Volts)'].to_list()) # intensity
    # errorbars
    x_err = 0.00001
    y_err = np.abs(-np.sin(2*x)*x_err)
    #print(y_err[0:5])

    # horizontal line
    ax.axhline(y=0, color="orange")
    # curve of best fit
    fit = a[i]*x
    res = y-fit
    
    # plot
    ax.scatter(x, res)
    ax.set_xlabel("cos²(θ)", fontsize=15)
    ax.set_ylabel("Light Intensity Residuals (V)", fontsize=15)
    ax.errorbar(x, res, xerr=x_err, yerr=y_err, fmt='none')
    fig.set_size_inches(10, 7)

    # save
    #ax.legend()
    name = "./intcos2/intcos2_res_" + str(i) + ".png"
    fig.savefig(name, bbox_inches='tight', pad_inches=0.2)

    i += 1

print(a)
```

```python
# parameter unc
# only doing it for 
a_unc = [0] * len(ex1)

for i in range(0, len(ex1)):

    # sort df so y aligns with fit data
    df = ex1[i].sort_values(by=['Sensor Position (Radians)'])
    # make the data
    x = np.array((np.cos(ex1[i]['Sensor Position (Radians)'].to_list()))**2) # cos(θ)
    y = np.array(ex1[i]['Light Intensity (Volts)'].to_list()) # intensity
    # errorbars
    x_err = 0.00001
    y_err = np.abs(-np.sin(2*x)*x_err)

    N = len(x)
    # delta
    delta = N*np.sum(x**2) - np.sum(x)**2
    #print("delta:", delta)
    # covariance
    cov = (1/(N-2)) * np.sum((y-a[i]*x)**2)
    #print("cov:", cov)
    
    # a_unc
    a_unc[i] = np.sqrt((N/delta) * cov)
    #print()
    
print(a_unc)
```

```python
# chi-squared
chisq = [0] * len(ex1)
rchisq = [0] * len(ex1) # reduced chi-squared

for i in range(0, len(ex1)):

    # sort df so y aligns with fit data
    df = ex1[i].sort_values(by=['Sensor Position (Radians)'])
    # make the data
    x = np.array((np.cos(ex1[i]['Sensor Position (Radians)'].to_list()))**2) # cos(θ)
    y = np.array(ex1[i]['Light Intensity (Volts)'].to_list()) # intensity
    # errorbars
    x_err = 0.00001
    y_err = np.abs(-np.sin(2*x)*x_err)
    
    N = len(x)
    # fit
    f = malus(x, a[i])

    # chi-squared
    chisq[i] = np.sum( (y-f)**2/np.var(y) )
    # reduced chi-squared
    rchisq[i] = chisq[i]/(N-1)

print(chisq)
print(rchisq)
```

```python
# mean and sd of values
print("I0")
print(a)
print(np.mean(a))
print(np.sqrt(np.var(a)))
print()

print("reduced χ2")
print(rchisq)
print(np.mean(rchisq))
print(np.sqrt(np.var(rchisq)))
print()
```

### Intensity vs (sin(2phi))^2

```python
# where x is sin²(2ϕ)
# a is I_1
def malus3(x, a):
    return a/4 * x
```

```python
a = [0] * len(ex2)

for i in range(0, len(ex2)):
    # make the data
    x = np.array((np.sin(2 * (ex2[i]['Sensor Position (Radians)']+np.pi/4)))**2) # sin(2phi)**2
    y = np.array(ex2[i]['Light Intensity (Volts)'].to_list()) # intensity
    
    # errorbars
    x_err = 0.00001
    y_err = 2*np.sin(4 * (ex2[i]['Sensor Position (Radians)'] + np.pi/4)) *x_err
    #print(y_err[0:5])
    
    # plot
    fig, ax = plt.subplots()
    ax.scatter(x, y, label='data')
    ax.set_ylabel("Light Intensity (V)", fontsize=15)
    ax.set_xlabel("sin²(2ϕ)", fontsize=15)
    #ax.errorbar(x, y, xerr=x_err, yerr=y_err[i], fmt='none')
    fig.set_size_inches(10, 7)

    # fit
    popt, pcov = curve_fit(malus3, x, y, p0 = (4))
    ax.plot(x, malus3(x, *popt), label='Malus\' Law (linear fit)', color='orange')
    a[i] = popt[0] # save for later
    print(popt)
    
    # save
    ax.legend()
    name = "./3pol/intsin2_" + str(i) + ".png"
    fig.savefig(name, bbox_inches='tight', pad_inches=0.5)

print(a)
```

```python
# residuals
for i in range(0, len(ex2)):

    fig, ax = plt.subplots()

    # sort df so y aligns with fit data
    df = ex2[i].sort_values(by=['Sensor Position (Radians)'])

    # make the data
    x = np.array((np.sin(2 * (ex2[i]['Sensor Position (Radians)']+np.pi/4)))**2) # sin(2phi)**2
    y = np.array(ex2[i]['Light Intensity (Volts)'].to_list()) # intensity
    # errorbars
    x_err = 0.00001
    y_err = np.abs(2*np.sin(4 * (ex2[i]['Sensor Position (Radians)'] + np.pi/4)) *x_err)
    #print(y_err[0:5])

    # horizontal line
    ax.axhline(y=0, color="orange")
    # curve of best fit
    fit = malus3(x, a[i])
    res = y-fit
    
    # plot
    ax.scatter(x, res)
    ax.set_xlabel("sin²(2ϕ)", fontsize=15)
    ax.set_ylabel("Light Intensity Residuals (V)", fontsize=15)
    ax.errorbar(x, res, xerr=x_err, yerr=y_err, fmt='none')
    fig.set_size_inches(10, 7)

    # save
    #ax.legend()
    name = "./3pol/intsin2_res_" + str(i) + ".png"
    fig.savefig(name, bbox_inches='tight', pad_inches=0.2)

    i += 1

print(a)
```

```python
# parameter unc
a_unc = [0] * len(ex2)

for i in range(0, len(ex2)):

    # sort df so y aligns with fit data
    df = ex2[i].sort_values(by=['Sensor Position (Radians)'])
    # make the data
    x = np.array((np.sin(2 * (ex2[i]['Sensor Position (Radians)']+np.pi/4)))**2) # sin(2phi)**2
    y = np.array(ex2[i]['Light Intensity (Volts)'].to_list()) # intensity
    # errorbars
    x_err = 0.00001
    y_err = np.abs(2*np.sin(4 * (ex2[i]['Sensor Position (Radians)'] + np.pi/4)) *x_err)

    N = len(x)
    # delta
    delta = N*np.sum(x**2) - np.sum(x)**2
    #print("delta:", delta)
    # covariance
    cov = (1/(N-2)) * np.sum((y-a[i]*x)**2)
    #print("cov:", cov)
    
    # a_unc
    a_unc[i] = np.sqrt((N/delta) * cov)
    #print()
    
print(a_unc)
```

```python
# chi-squared
chisq = [0] * len(ex2)
rchisq = [0] * len(ex2) # reduced chi-squared

for i in range(0, len(ex2)):

    # sort df so y aligns with fit data
    df = ex2[i].sort_values(by=['Sensor Position (Radians)'])
    # make the data
    x = np.array((np.sin(2 * (ex2[i]['Sensor Position (Radians)']+np.pi/4)))**2) # sin(2phi)**2
    y = np.array(ex2[i]['Light Intensity (Volts)'].to_list()) # intensity
    # errorbars
    x_err = 0.00001
    y_err = np.abs(2*np.sin(4 * (ex2[i]['Sensor Position (Radians)'] + np.pi/4)) *x_err)
    
    N = len(x)
    # fit
    f = malus(x, a[i])

    # chi-squared
    chisq[i] = np.sum( (y-f)**2/np.var(y) )
    # reduced chi-squared
    rchisq[i] = chisq[i]/(N-1)

print(chisq)
print(rchisq)
```
